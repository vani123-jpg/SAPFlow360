# READ ALL 8 DATASETS

```python
import pandas as pd                 # Data read, cleaning and analysis-ku use panrom
import numpy as np                  # Calculations-ku use panrom
import matplotlib.pyplot as plt     # Basic chart support-ku use panrom
import seaborn as sns               # Main visualizations-ku use panrom

print("Imports successful")
# All required libraries correct-ah load aachaa-nu check panrom.
```

```python

vendors = pd.read_csv("data/vendors.csv", low_memory=False)
# Dataset 1 → Vendors
# Company-ku materials supply panra suppliers-oda information.
# low_memory=False → Large/mixed data irundhaalum Pandas safer-ah read panna help pannum.


purchase_orders = pd.read_csv(
    "data/purchase_orders.csv",
    low_memory=False
)
# Dataset 2 → Purchase Orders
# Company suppliers-kitta enna orders place pannuchu-nu information.
# Idhu namma main purchasing/order-level data.


poi = pd.read_csv(
    "data/purchase_order_items.csv",
    low_memory=False
)
# Dataset 3 → PO Items
# Andha Purchase Orders-kulla exactly enna items irukku-nu information.
# Material, quantity, price, value maari details irukkum.


materials = pd.read_csv(
    "data/materials.csv",
    low_memory=False
)
# Dataset 4 → Materials
# Purchase pannra materials-oda master details.
# Material type, group, unit, created date maari information.


names = pd.read_csv(
    "data/material_names.csv"
)
# Dataset 5 → Material Names
# Material ID-ku actual material name/description provide pannum.
# Example: Material ID → M001, Material Name → Steel.


inventory = pd.read_csv(
    "data/inventory.csv",
    low_memory=False
)
# Dataset 6 → Inventory
# Materials-ku available stock evlo irukku-nu information.
# Stock tracking and current inventory analysis-ku use pannuvom.


plants = pd.read_csv(
    "data/plants.csv"
)
# Dataset 7 → Plants
# Company-oda stock/material endha plant or location-ku belong aagudhu-nu information.


groups = pd.read_csv(
    "data/material_groups.csv"
)
# Dataset 8 → Material Groups
# Material endha group/category-ku belong aagudhu-nu information.


print("All 8 datasets loaded successfully.")
# All 8 datasets proper-ah load aachaa-nu simple confirmation message.
```

# EXPLORE ALL 8 DATASETS

“8 datasets-ku same same exploration code repeat-aa ezhutha vendam. Dictionary-la 8 datasets-a store panni, for loop use panni one by one explore panrom. Idhu time save pannum.”

1. name = dataset name
2. df = dataset
3. for = one by one

```python



datasets = {
    "Vendors": vendors,
    "Purchase Orders": purchase_orders,
    "PO Items": poi,
    "Materials": materials,
    "Material Names": names,
    "Inventory": inventory,
    "Plants": plants,
    "Material Groups": groups
}
# All 8 datasets-a one dictionary-la store panrom.
# Later same exploration steps-a all datasets-kum apply panna easy-ah irukkum.


for name, df in datasets.items():


    print("Shape:", df.shape)
    # Ethana rows + columns irukku-nu check panrom.

    print("\nFirst 3 rows:")
    print(df.head(3))
    # Starting-la data epdi irukku-nu paakrom.

    print("\nColumns:")
    print(df.columns.tolist())
    # Enna information available-nu check panrom.

    print("\nInfo:")
    df.info()
    # Data types and non-null values-a check panrom.

    print("\nMissing values:")
    print(
        df.isna().sum()[
            df.isna().sum() > 0
        ]
    )
    # Missing values irukkura columns mattum kaamikrom.

    print("\nDuplicate rows:")
    print(df.duplicated().sum())
    # Exact duplicate rows ethana-nu check panrom.
```
