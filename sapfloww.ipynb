# dropna :
Without dropna=False, Pandas normally doesn't show the missing values.
dropna=False means: include the missing (NaN) values when counting.

# Filter / Current-state selection

```python
inventory["recordstamp"] = pd.to_datetime(
    inventory["recordstamp"],
    format="mixed",
    utc=True
)

# format="mixed" = tells Pandas that the dates may be written in different formats. 
#  eg. 2022-01-15, 01/15/2022, 2022/01/15

#UTC = a common worldwide time standard used to keep date and time consistent.
#utc=True → keep all times in one standard
```

Then keep the latest record for each:

Material + Plant + Storage Location

```python
inventory_current = (
    inventory
    .sort_values("recordstamp") 
    .groupby(
        [
            "matnr",
            "werks",
            "lgort"
        ],
        as_index=False
    )
    .tail(1)
)

#Sorts the records according to their date and time, from oldest to newest,
#Arrange the data in order.
#
```

1. groupby(["matnr", "werks", "lgort"])
# matnr → Material
# werks → Plant
# lgort → Storage Location

For each material, at each plant, and at each storage location, treat those records as one group

Material A | Plant 1 | Location X
Material A | Plant 1 | Location X
Material A | Plant 1 | Location X

# These belong to the same group.

# as_index=False 
Keeps matnr, werks, and lgort as normal columns instead of turning them into an index. 

# Clean

```python
inventory_mp = (
    inventory_current
    .groupby(
        [
            "matnr",
            "werks"
        ],
        as_index=False
    )["labst"]
    .sum(min_count=1)
)
print(inventory_mp)
```

Company-oda material/stock endha location that factory present it means that.
Example:

Plant P01 → Chennai
Plant P02 → Bangalore


1. Groups the stock by Material and Plant.

“For each material, how much stock is available at each plant?”
Example:

Material A | Plant 1 | 100 KG
Material A | Plant 1 | 50 KG
Material A | Plant 2 | 200 KG

After grouping:

Material A | Plant 1 → 150 KG
Material A | Plant 2 → 200 KG

# .sum(min_count=1) 
→ Adds the available stock for each Material + Plant combination.
# one more change (NuN)
possibility- Because maybe the stock was not recorded, not available, or missing.

prevents Pandas from turning a group containing only missing values into 0
Example:

Material A + Plant 1
100
50
↓
150





# Validate

```python
print(
    "Current inventory rows:",
    len(inventory_current)
)

print(
    "Current rows with stock:",
    inventory_current["labst"].notna().sum()
)

print(
    "Current rows without stock:",
    inventory_current["labst"].isna().sum()
)

print(
    "Duplicate Material + Plant + Location:",
    inventory_current.duplicated(
        [
            "matnr",
            "werks",
            "lgort"
        ]
    ).sum()
)

print(
    "Material + Plant rows:",
    len(inventory_mp)
)

print(
    "Missing stock:",
    inventory_mp["labst"].isna().sum()
)
```

```python

# CHART 7 — PURCHASE QUANTITY vs PURCHASE VALUE


g = sns.jointplot(
    data=poi_clean,
    x="menge",
    y="netwr",
    kind="reg",                 # Quantity and Value-kulla overall trend-a show panrom
    height=9,                   # Overall chart-a bigger-ah make panrom
    ratio=6,                    # Main chart-ku more space kudukrom
    space=0.15,                 # Main chart-um side charts-um konjam gap vechukrom
    scatter_kws={
        "alpha": 0.20,          # Too many points overlap aagama slightly transparent-ah vechukrom
        "s": 18                 # Individual points small-ah vechukrom
    },
    line_kws={
        "linewidth": 2.5        # Trend line-a clear-ah theriyura maari thick panrom
    }
)

# Very large values irukkumbodhu small and large values rendu-m better-ah paaka
# X and Y axes-a logarithmic scale-ku maathrom.
g.ax_joint.set_xscale("log")
g.ax_joint.set_yscale("log")

# Correlation calculate panrom
correlation = (
    poi_clean["menge"]
    .corr(poi_clean["netwr"])
)

# Correlation value-a chart-la highlight panrom
g.ax_joint.text(
    0.05,
    0.92,
    f"Correlation: {correlation:.2f}",
    transform=g.ax_joint.transAxes,
    fontsize=11,
    fontweight="bold",
    bbox=dict(
        boxstyle="round,pad=0.4",
        alpha=0.9
    )
)

g.set_axis_labels(
    "Purchase Quantity",
    "Purchase Value",
    fontsize=11
)
# X-axis → Purchase Quantity
# Y-axis → Purchase Value

g.ax_joint.set_title(
    "Purchase Quantity vs Purchase Value",
    fontsize=16,
    fontweight="bold",
    pad=20
)
# Chart-ku clear and attractive title kudukrom.

plt.show()
# Final chart-a display panrom.
```
